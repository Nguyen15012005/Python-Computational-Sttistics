### **Bài 2**
Lấy ngẫu nhiên 2000 mẫu dữ liệu (sample = data.sample(n, random_state=11)) từ dữ liệu gốc, có nhận xét cho rằng: "Có từ 98% thí sinh thi khối KHTN có điểm Văn trên trung bình (>=5)." Bạn có thể đưa ra kết luận gì nhận xét trên dựa vào 2000 mẫu vừa lấy được với mức ý nghĩa 5%? (4đ)

```python
# 0.25đ
```

Gọi $p$ là thí sinh có điểm Văn trên trung bình. Ta cần kiểm định giả thuyết:

$H_0 : p \ge 0.98$: Có từ 98% thí sinh khối KHTN có điểm Văn trên trung bình **(0.5đ)**

$H_1: p < 0.98$: Tỉ lệ thí sinh khối KHTN có điểm văn trên trung bình thấp hơn 98%$ **(0.5đ)**

Mức ý nghĩa là: $5\% => \alpha=0.05$ **(0.25đ)**

- Điểm thi của các thí sinh trên trung bình hay không là độc lập với nhau nên tỉ lệ thí sinh đạt điểm trên trung bình tuân theo phân phối nhị thức. **(0.25đ)**
- Ta có
    - $\hat{p} = 0.98$ là tỷ lệ thí sinh có điểm văn trên trung bình.
    - $n=2000 > 30$, **(0.25đ)**
    - $\hat{p}=0.98$, $n\hat{p}=2000\times0.98=1960 \geq 5$ **(0.25đ)**
    - $n(1-\hat{p}) = 2000(1-0.98)= 40 \geq 5$ **(0.25đ)**
    
=> Vậy phân phối nhị thức của tỷ lệ mẫu có thể xấp xỉ bằng phân phối chuẩn. **(0.25đ)**
$$z=\frac{\hat{p}-p}{\sqrt{\frac{pq}{n}}}$$

```python
# 0.25đ
```

**(0.25đ)**
Vì $H_1$ là $<$ là kiểm định đuôi trái nên ta có
$$p_{values} = P(Z<z)$$

```python
# 0.25đ
```

**(0.5đ)** Vì $p_{value} = 0.0007 < \alpha=0.05$ nên ta kết luận: Nhận xét trên là không đúng với mức ý nghĩa 5%$
