```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data= pd.read_csv("StudentSurveyGK.csv")
#Cau 1
data.head()

#Cau 2a
piercings_percent= (data['Piercings']>0).value_counts(normalize=True)*100
print("Ti le sinh vien deo khuyen: ",round(piercings_percent.iloc[1],2))
#Cau 2b
male_piercings_percent= (data[data['Gender']=='M']['Piercings']>0).value_counts(normalize=True)*100
print("Ti le sinh vien nam deo khuyen: ",round(male_piercings_percent.iloc[1],2))

female_piercings_percent= (data[data['Gender']=='F']['Piercings']>0).value_counts(normalize=True)*100
print("Ti le sinh vien nu deo khuyen: ",round(female_piercings_percent.iloc[0],2))

print("Vay ti le nu deo khuyen cao hon ty le nam deo khuyen")
#Cau 3
plt.hist(data['Pulse'], bins=10, edgecolor='black')
plt.xlabel('Pulse')
plt.ylabel('Frequency')
plt.title('Histogram of Pulse')
plt.show()

plt.hist(data['Exercise'], bins=10, edgecolor='black')
plt.xlabel('Exercise')
plt.ylabel('Frequency')
plt.title('Histogram of Exercise')
plt.show()

plt.hist(data['Piercings'], bins=10, edgecolor='black')
plt.xlabel('Piercings')
plt.ylabel('Frequency')
plt.title('Histogram of Piercings')
plt.show()

#Phan phoi thuoc tinh Pulse doi xung
#Phan phoi thuoc tinh Exercise lech trai
#Phan phoi thuoc tinh Piercings lech trai

#Cau 4a
sample_4a= data.sample(n=50, random_state=1)
female_percent_in_sample= ((sample_4a['Gender']=='F').value_counts(normalize=True)*100).iloc[1]
print("Ti le sinh vien nu trong sample: ",round(female_percent_in_sample,2))
female_percent= ((data['Gender']=='F').value_counts(normalize=True)*100).iloc[1]
print("Ti le sinh vien nu trong toan bo du lieu: ",round(female_percent,2))

print("Sai so: ", round(female_percent-female_percent_in_sample,2))

#Cau 4b
female_percent_in_sample_4b= []
for i in range(100):
  female_percent_4b= ((data.sample(n=50)['Gender']=='F').value_counts(normalize=True)*100).iloc[1]
  female_percent_in_sample_4b.append(female_percent_4b)
plt.hist(female_percent_in_sample_4b, bins=10, edgecolor='black')
plt.xlabel('Female Percent')
plt.ylabel('Frequency')
plt.title('Histogram of Female Percent')
plt.show()

#Cau 4c
female_percent_in_sample_4c1= []
for i in range(100):
  female_percent_4b= ((data.sample(n=100)['Gender']=='F').value_counts(normalize=True)*100).iloc[1]
  female_percent_in_sample_4c1.append(female_percent_4b)
plt.hist(female_percent_in_sample_4c1, bins=10, edgecolor='black')
plt.xlabel('Female Percent')
plt.ylabel('Frequency')
plt.title('Histogram of Female Percent 1')
plt.show()

female_percent_in_sample_4c2= []
for i in range(100):
  female_percent_4b= ((data.sample(n=200)['Gender']=='F').value_counts(normalize=True)*100).iloc[1]
  female_percent_in_sample_4c2.append(female_percent_4b)
plt.hist(female_percent_in_sample_4c2, bins=10, edgecolor='black')
plt.xlabel('Female Percent')
plt.ylabel('Frequency')
plt.title('Histogram of Female Percent 2')
plt.show()
```
