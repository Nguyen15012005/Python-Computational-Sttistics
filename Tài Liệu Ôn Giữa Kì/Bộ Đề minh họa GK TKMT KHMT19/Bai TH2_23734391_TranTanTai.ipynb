Câu 1. Một đề thi trắc nghiệm gồm có 40 câu hỏi, mỗi câu hỏi có 4 phương án trả lời, trong đó chỉ có 01 phương án trả lời đúng. 
Một thí sinh làm bài bằng cách chọn ngẫu nhiên các phương án trả lời cho tất cả các câu hỏi.

a. (1 điểm) Tính xác suất thí sinh trả lời đúng 20 câu hỏi.

b. (2 điểm) Mỗi câu trả lời đúng được 0.25 điểm. Tính xác suất thí sinh đạt được từ 4 điểm trở lên.

```python
from scipy.stats import binom

n = 40  
p = 0.25  
# a
k = 20
prob_a = binom.pmf(k, n, p)
print(f"Xác suất trả lời đúng 20 câu hỏi: {prob_a:.10f}")
```

```python
# b
k_min = 16
prob_b = binom.sf(k_min - 1, n, p)
print(f"Xác suất đạt được từ 4 điểm trở lên: {prob_b:.10f}")
```

Câu 2. Để đánh giá kết quả học tập cuối năm của học sinh lớp 12 của tỉnh A, người ta chọn ngẫu nhiên 40 học sinh, điểm tổng kết của các học sinh như sau:
$5.1, 5.5, 7.6, 8.0, 8.5, 6.5, 6.6, 7.2, 4.8, 7.5, 8.8, 5.7, 6.9, 8.2, 7.7, 6.3, 5.6, 7.0, 7.4, 7.3, 6.9, 5.7, 5.3, 6.0, 5.0, 8.4, 9.2, 7.4, 7.9, 5.9, 
6.6, 7.8, 8.7, 8.8, 6.8, 7.1, 7.5, 6.8, 5.8, 6.4$

a. (3 điểm) Tìm khoảng ước lượng điểm tổng kết trung bình của học sinh lớp 12 của tỉnh A với độ tin cậy 90%, 95%, 99%.

b. (1 điểm) Tìm khoảng ước lượng phương sai điểm tổng kết của học sinh lớp 12 của tỉnh A với độ tin cậy 95%.

c. (1 điểm) Những học sinh có điểm trung bình từ 8.0 trở lên là học sinh giỏi. Hãy ước lượng tỷ lệ học sinh giỏi của tỉnh A với độ tin cậy 99%.

d. (2 điểm) Một báo cáo công bố rằng, tỷ lệ học sinh giỏi lớp 12 của tỉnh A là 27%. Với mức ỹ nghĩa 1%, báo cáo trên có đáng tin không?

```python
import numpy as np
from scipy.stats import t, chi2

data = np.array([5.1, 5.5, 7.6, 8.0, 8.5, 6.5, 6.6, 7.2, 4.8, 7.5, 8.8, 5.7, 6.9, 8.2, 7.7, 6.3, 5.6, 7.0, 7.4,
                 7.3, 6.9, 5.7, 5.3, 6.0, 5.0, 8.4, 9.2, 7.4, 7.9, 5.9, 6.6, 7.8, 8.7, 8.8, 6.8, 7.1, 7.5, 6.8, 5.8, 6.4])
n = len(data)
mean = np.mean(data)
std_dev = np.std(data, ddof=1)

# a
def uocluongdiemtongkettrungbinh(a):
    alpha = 1 - a
    t_value = t.ppf(1 - alpha / 2, n - 1)
    margin_of_error = t_value * std_dev / np.sqrt(n)
    return mean - margin_of_error, mean + margin_of_error

ci_90 = uocluongdiemtongkettrungbinh(0.90)
ci_95 = uocluongdiemtongkettrungbinh(0.95)
ci_99 = uocluongdiemtongkettrungbinh(0.99)

print("Khoảng ước lượng điểm trung bình:")
print(f"90%: {ci_90}")
print(f"95%: {ci_95}")
print(f"99%: {ci_99}")
```

```python
# b
def uocluongphuongsai(b):
    alpha = 1 - b
    chi2_lower = chi2.ppf(alpha / 2, n - 1)
    chi2_upper = chi2.ppf(1 - alpha / 2, n - 1)
    variance = std_dev ** 2
    lower_bound = (n - 1) * variance / chi2_upper
    upper_bound = (n - 1) * variance / chi2_lower
    return lower_bound, upper_bound

ci_variance_95 = uocluongphuongsai(0.95)

print("Khoảng ước lượng phương sai:")
print(f"95%: {ci_variance_95}")
```

```python
# c
excellent_students = len(data[data >= 8.0])
excellent_ratio = excellent_students / n

def uocluonghocsinhgioi(c, p):
    alpha = 1 - c
    z_value = t.ppf(1 - alpha / 2, n - 1)
    margin_of_error = z_value * np.sqrt(p * (1 - p) / n)
    return p - margin_of_error, p + margin_of_error

ci_proportion_99 = uocluonghocsinhgioi(0.99, excellent_ratio)

print("Khoảng ước lượng tỷ lệ học sinh giỏi:")
print(f"99%: {ci_proportion_99}")
```

```python
# d
p0 = 0.27
z_score = ((excellent_ratio - p0) / np.sqrt(p0 * (1 - p0) / n)).round(4)
alpha = 0.01
critical_value = t.ppf(1 - alpha / 2, n - 1).round(4)
print("z_score:",z_score)
print("critical_value:", critical_value)
if abs(z_score) > critical_value:
    print("Bác bỏ giả thuyết tỷ lệ học sinh giỏi là 27%")
else:
    print("Không thể bác bỏ giả thuyết tỷ lệ học sinh giỏi là 27%")
```
