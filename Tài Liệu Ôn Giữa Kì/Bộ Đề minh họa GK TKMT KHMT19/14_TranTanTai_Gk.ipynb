```python
#Cau 1a
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('IT-salaries-in-EU-2018.csv')

# Hiển thị các thuộc tính của dữ liệu
print("Thông tin về dữ liệu:")
print(df.info())

print("\nNăm dòng đầu tiên của dữ liệu:")
print(df.head())

print("\nCác cột trong dữ liệu:")
print(df.columns)

print("\nHình dạng của dữ liệu (số hàng, số cột):")
print(df.shape)

print("\nCác kiểu dữ liệu trong dữ liệu:")
print(df.dtypes)
```

```python
#Cau1b

data = df.dropna(how='all', axis=1)

print("Thông tin tổng quan về dữ liệu:")
print(df.info())

qualitative = []  
quantitative = []  

for col in data.columns:
    if df[col].dtype == 'object':
        qualitative.append(col)
    else:
        quantitative.append(col)

print("\nThuộc tính định tính:")
print(qualitative)

print("\nThuộc tính định lượng:")
print(quantitative)
if quantitative:
    print("\nThống kê mô tả cho các thuộc tính định lượng:")
    print(df[quantitative].describe())
```

```python
#Cau1c

company_types = df['Company type'].dropna()


unique_company_types = company_types.unique()
count_company_types = len(unique_company_types)

print(f"Số loại công ty: {count_company_types}")
print("Danh sách các loại công ty:")
for company in unique_company_types:
    print("-", company)
```

```python
#Cau1d
data = df.dropna(subset=['Gender', 'Current Salary'])
grouped_data = df.groupby('Gender')['Current Salary']

mean_salary = grouped_data.mean()
median_salary = grouped_data.median()
std_salary = grouped_data.std()

print("Mức lương trung bình theo giới tính:")
print(mean_salary)
print("\nMức lương trung vị theo giới tính:")
print(median_salary)
print("\nĐộ lệch chuẩn về mức lương theo giới tính:")
print(std_salary)

plt.figure(figsize=(10, 6))
data.boxplot(column='Current Salary', by='Gender', grid=False, showmeans=True)
plt.title('So sánh mức lương hiện tại giữa Nam và Nữ')
plt.suptitle("")
plt.xlabel('Giới tính')
plt.ylabel('Mức lương hiện tại')
plt.show()
```

```python
#Cau2a
import numpy as np
import scipy.stats as stats
import pandas as pd
P1 = stats.norm.cdf(1.96)
P2 = 1 - stats.norm.cdf(2.57)
print(f"P(X < 1.65): {P1}")
print(f"P(X > 2.57): {P2}")
```

```python
#Cau3 
data = df.dropna(subset=['Current Salary'])

current_salaries = data['Current Salary']


mean_salary = np.mean(current_salaries)
std_salary = np.std(current_salaries, ddof=1)

n = len(current_salaries)

confidence_interval = stats.t.interval(0.95, df=n-1, loc=mean_salary, scale=std_salary/np.sqrt(n))

print("Mức lương trung bình hiện tại (ước lượng):", mean_salary)
print("Khoảng tin cậy 95%:", confidence_interval)
```
