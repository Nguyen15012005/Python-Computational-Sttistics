### Câu 1.2

Liệt kê tên tất cả các cột. Phân loại mỗi cột thuộc kiểu dữ liệu nào: **định lượng liên tục**, **định lượng rời rạc**, **định tính (danh nghĩa)**, hay **định tính (thứ tự)**?

```python
# Câu 1.2

# Kiểu Pandas	Phân loại thống kê
# float	Định lượng liên tục
# int	Định lượng rời rạc
# object / string	Định tính (danh nghĩa)
# category (ordered)	Định tính (thứ tự)

# Liệt kê tất cả các cột
print(df.columns.tolist())

# Xem kiểu dữ liệu gốc
print(df.dtypes)

# Phân loại mỗi cột dữ liệu theo định dạng:
def classify_column(col):
    name = col.name.lower()
    dtype = col.dtype
    
    if "mã" in name or "id" in name:
        return "Định tính (danh nghĩa)"
    if "đánh giá" in name:
        return "Định tính (thứ tự)"
    if "mã" in name or "id" in name:
        return "Định tính (danh nghĩa)"
    if "trạng thái" in name:
        return "Định tính (danh nghĩa)"
    if "ngày" in name:
        return "Định tính (thứ tự)"
    
    if pd.api.types.is_float_dtype(dtype):
        return "Định lượng liên tục"
    
    elif pd.api.types.is_integer_dtype(dtype):
        return "Định lượng rời rạc"
    
    else:
        return "Định tính danh nghĩa"

# In ra kết quả
for col in df.columns:
    print(f"{col}                 : {classify_column(df[col])}")
    
result = pd.DataFrame({
    "Tên cột": df.columns,
    "Phân loại": [classify_column(df[col]) for col in df.columns]
})

print(result)
```

### Câu 1.9

Kiểm tra tính nhất quán dữ liệu: 

(a) Có dòng nào `Thành tiền (VNĐ)` > `Doanh thu (VNĐ)` không? 

(b) Có đơn hàng nào `Giảm giá (%)` > 0 nhưng `Thành tiền` = `Doanh thu` không? 

(c) Kiểm tra công thức `Thành tiền = Doanh thu × (1 - Giảm giá/100)` có đúng cho tất cả các dòng không (chấp nhận sai số làm tròn ≤ 1000 VNĐ)?

```python
# Câu 1.9
# a
df_a = df[df["Thành tiền (VNĐ)"] > df["Doanh thu (VNĐ)"]]
print(df_a)

# b
df_b = df[
            (df["Giảm giá (%)"] > 0) &
            (df["Thành tiền (VNĐ)"] == df["Doanh thu (VNĐ)"])
         ]
print(df_b)

# c
df["Thanh_tien_calc"] = df["Doanh thu (VNĐ)"]*(1 - df["Giảm giá (%)"]/100)

# Tính sai lệch
df["Sai_lech"] = df["Thành tiền (VNĐ)"] - df["Thanh_tien_calc"]

df_c = df[df["Sai_lech"] > 1000]
print(df_c)
```

Nhận xét Câu 1.9:

Câu a:
- Không phát hiện dòng nào có Thành tiền lớn hơn Doanh thu → dữ liệu hợp lý.

Câu b:
- Phát hiện 1 đơn hàng có Giảm giá > 0 nhưng Thành tiền = Doanh thu → dấu hiệu dữ liệu không nhất quán, có thể do lỗi nhập liệu hoặc tính toán.

Câu c:
- Công thức Thành tiền = Doanh thu × (1 - Giảm giá/100) đúng cho tất cả các dòng trong phạm vi sai số làm tròn ≤ 1000 VNĐ.

Tạo một **Data Profile Report** tóm tắt toàn bộ bộ dữ liệu bao gồm: số dòng, số cột, số cột định lượng vs định tính, tổng missing values, tỷ lệ missing, số dòng trùng lặp (duplicate rows), dung lượng bộ nhớ (memory usage). 

Trình bày kết quả dưới dạng bảng gọn gàng.

```python
# Câu 1.10
# lấy ra số dòng số cột
num_rows, num_cols = df.shape

# Lấy ra số cột định lượng và định tính
# Numeric: giá, số lượng, tuổi
# Categorical: khu vực, sản phẩm
num_cols_numeric = df.select_dtypes(include=["number"]).shape[1]
num_cols_categorical = df.select_dtypes(exclude=["number"]).shape[1]

# tổng missing values
total_missing = df.isnull().sum().sum()

# Tính tỷ lệ missing
total_rows_cols = num_rows * num_cols
missing_ratio = (total_missing / total_rows_cols * 100)

# số dòng trùng lặp (duplicate rows)
duplicate_rows = df.duplicated().sum()

# dung lượng bộ nhớ (memory usage)
memory_usage = df.memory_usage(deep=True).sum() / (1024*1024)

report = pd.DataFrame({
    "Chỉ số": [
        "Số dòng",
        "Số cột",
        "Số cột định lượng",
        "Số cột định tính",
        "Tổng missing values",
        "Tỷ lệ missing (%)",
        "Số dòng trùng lặp",
        "Dung lượng bộ nhớ"
    ],
    "Giá trị": [
        num_rows, 
        num_cols,
        num_cols_numeric,
        num_cols_categorical,
        total_missing,
        round(missing_ratio, 2),
        duplicate_rows,
        round(memory_usage, 2)
    ]
})

report
```

### Câu 2.4

Tính **Q1, Q2 (median), Q3, IQR** của cột `Số lượng`. Xác định ngưỡng outlier trên và dưới theo phương pháp IQR.

```python
# Câu 2.4
# Tính Q1, 2, 3 IQR
Q1 = df["Số lượng"].quantile(0.25)
Q2 = df["Số lượng"].median()
Q3 = df["Số lượng"].quantile(0.75)
print("Q1",Q1)
print("Q2",Q2)
print("Q3",Q3)
IQR = Q3 - Q1
print("IQR",IQR)

# Tính ngưỡng outlier
lower_bound = Q1 - 1.5*IQR
upper_bound = Q3 + 1.5*IQR
print("lower_bound",lower_bound)
print("upper_bound",upper_bound)

outliers = df[(df["Số lượng"] < lower_bound) | (df["Số lượng"] > upper_bound)]

print(outliers)
# 
```

### Câu 2.6

Tính **hệ số biến thiên CV** cho 4 cột: `Đơn giá (VNĐ)`, `Số lượng`, `Tuổi KH`, `Phí vận chuyển (VNĐ)`. Biến nào phân tán nhất? Biến nào tập trung nhất?

```python
# Câu 2.6
cols = ["Đơn giá (VNĐ)", "Số lượng", "Tuổi KH", "Phí vận chuyển (VNĐ)"]
cv = {}

for col in cols:
    mean = df[col].mean()
    std = df[col].std()
    cv[col] = std / mean
    
cv

cv_df = pd.DataFrame(list(cv.items()), columns=["Biến", "CV"])
cv_df = cv_df.sort_values(by="CV", ascending=False)

cv_df
# Nhận xét
# Biến số lượng phân tán nhất
# Biến tuổi KH tập trung nhất
```

### Câu 2.8

Tính **doanh thu trung bình mỗi ngày** (tổng `Thành tiền (VNĐ)` / số ngày). 

Tính **doanh thu trung bình mỗi đơn hàng**. 

Tính **số đơn hàng trung bình mỗi ngày**.

```python
# Câu 2.8
# df["Ngày đặt hàng"] = pd.to_datetime(df["Ngày đặt hàng"])
total_revenue = df["Thành tiền (VNĐ)"].sum()

num_days = df["Ngày đặt hàng"].nunique()

avg_revenue_per_day = total_revenue / num_days

print("Doanh thu trung bình mỗi ngày",avg_revenue_per_day)

avg_revenue_per_other = df["Thành tiền (VNĐ)"].mean()
print("Doanh thu trung bình mỗi đơn hàng",avg_revenue_per_other)

total_orders = len(df)
avg_orders_per_day = total_orders / num_days

print("Số đơn hàng trung bình mỗi ngày", avg_orders_per_day)
```

### Câu 2.9

Hệ số tương quan dùng để làm gì?

Tính **hệ số tương quan Pearson** giữa các cặp biến: (Đơn giá, Số lượng), (Tuổi KH, Thành tiền), (Đánh giá, Thành tiền). 

Lập **ma trận tương quan** cho tất cả biến số. Nhận xét: cặp nào tương quan mạnh nhất? Có cặp nào tương quan âm không?

```python
# Câu 2.9
# Hệ số tương quan dùng để đo mức độ và chiều hướng mối quan hệ tuyến tính giữa hai biến.
# - Giá trị nằm trong khoảng [-1, 1]
# + r > 0 → tương quan thuận (tăng cùng nhau)
# + r < 0 → tương quan nghịch (một tăng, một giảm)
# + r ≈ 0 → không có mối quan hệ tuyến tính rõ ràng
# → Giá trị |r| càng lớn → mối quan hệ càng mạnh

# 1. Các cột cần dùng
cols = ["Đơn giá (VNĐ)", "Số lượng", "Tuổi KH", "Thành tiền (VNĐ)", "Đánh giá (sao)"]

# 2. Làm sạch dữ liệu (QUAN TRỌNG NHẤT)
for col in cols:
    df[col] = (
        df[col]
        .astype(str)
        .str.replace(",", "", regex=False)         # bỏ dấu phẩy
        .str.replace(r"[^\d.]", "", regex=True)    # bỏ ký tự chữ
    )
    df[col] = pd.to_numeric(df[col], errors="coerce")

# 3. Loại bỏ NaN
df_clean = df[cols].dropna()

# 4. Tính hệ số tương quan Pearson
corr_1 = df_clean["Đơn giá (VNĐ)"].corr(df_clean["Số lượng"])
corr_2 = df_clean["Tuổi KH"].corr(df_clean["Thành tiền (VNĐ)"])
corr_3 = df_clean["Đánh giá (sao)"].corr(df_clean["Thành tiền (VNĐ)"])

print("Đơn giá - Số lượng:", corr_1)
print("Tuổi KH - Thành tiền:", corr_2)
print("Đánh giá - Thành tiền:", corr_3)

# 5. Ma trận tương quan
corr_matrix = df_clean.corr()

print("\nMa trận tương quan:")
print(corr_matrix)

# 6. Tìm cặp tương quan mạnh nhất
corr_abs = corr_matrix.abs()
np.fill_diagonal(corr_abs.values, 0)

max_pair = corr_abs.unstack().idxmax()
max_value = corr_matrix.unstack()[max_pair]

print("\nCặp tương quan mạnh nhất:", max_pair)
print("Giá trị:", max_value)

# 7. Kiểm tra có tương quan âm không
if (corr_matrix < 0).any().any():
    print("\nCó tồn tại tương quan âm giữa các biến.")
else:
    print("\nKhông có tương quan âm đáng kể.")
```

### Câu 2.10

Tính **Percentile thứ 5, 10, 25, 50, 75, 90, 95** cho `Thành tiền (VNĐ)`. 

Dựa vào kết quả, hãy phân nhóm đơn hàng thành 4 nhóm: Nhỏ (≤P25), Trung bình (P25–P75), Lớn (P75–P95), Rất lớn (>P95). 

Đếm số đơn hàng và tính tổng doanh thu mỗi nhóm.

```python
# Câu 2.10

percentiles = df["Thành tiền (VNĐ)"].quantile([0.05, 0.1, 0.25, 0.5, 0.75, 0.9, 0.95])
print(percentiles)
p25 = percentiles.loc[0.25]
p75 = percentiles.loc[0.75]
p95 = percentiles.loc[0.95]
def classify(x):
    if x <= p25:
        return "Nhỏ"
    elif x <= p75:
        return "Trung bình"
    elif x <= p95:
        return "Lớn"
    else:
        return "Rất lớn"

df["Nhóm đơn hàng"] = df["Thành tiền (VNĐ)"].apply(classify)

summary = df.groupby("Nhóm đơn hàng")["Thành tiền (VNĐ)"].agg(
    so_don="count",
    tong_doanh_thu="sum"
)

print(summary)

```

```python
# Câu 3.8
# Tính IQR 
Q1 = df["Đơn giá (VNĐ)"].quantile(0.25)
Q3 = df["Đơn giá (VNĐ)"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5*IQR
upper_bound = Q3 + 1.5*IQR

outliers = df[
    (df["Đơn giá (VNĐ)"] < lower_bound) |
    (df["Đơn giá (VNĐ)"] > upper_bound) 
]

nhom_outliers = outliers["Nhóm sản phẩm"].value_counts()

print(nhom_outliers)
print("Nhóm nhiều outliers nhất: ", nhom_outliers.idxmax())
```

### Câu 3.9

Tìm **top 3 nhân viên** (theo `Mã nhân viên`) có tổng doanh thu (`Thành tiền`) cao nhất. Với mỗi nhân viên, cho biết: số đơn hàng, doanh thu trung bình/đơn, nhóm sản phẩm bán nhiều nhất, tỷ lệ đơn hoàn thành.

```python
# Câu 3.9
top3nv = df.groupby("Mã nhân viên")["Thành tiền (VNĐ)"].sum().sort_values(ascending=False).head(3)
print("3 nhân viên có tổng doanh thu (thành tiền) cao nhất\n",top3nv)
top3_ids = top3nv.index
df_top3 = df[df["Mã nhân viên"].isin(top3_ids)]

result = []

for nv in top3_ids:
    data_top3 = df_top3[df_top3["Mã nhân viên"] == nv]
    
#     Số đơn hàng
    so_don = len(data_top3)
    doanhthutb = data_top3["Thành tiền (VNĐ)"].mean()
    
#     Nhóm sản phẩm bán nhiều nhất
    nhom_sp = data_top3["Nhóm sản phẩm"].value_counts().idxmax()
    
#     Tỷ lệ đơn hoàn thành
    tyledonhoanthanh = ((data_top3["Trạng thái"] == "Hoàn thành").sum() / so_don) * 100
    
    result.append({
        "Mã nhân viên": nv,
        "Số đơn": so_don,
        "Doanh thu TB": doanhthutb,
        "Nhóm SP nhiều nhất": nhom_sp,
        "Tỷ lệ đơn hoàn thành %": tyledonhoanthanh
    })
    
ket_qua_df = pd.DataFrame(result)

print("Kết quả")
print(ket_qua_df)
```

```python
# Câu 4.8
# 1. Tạo cột Flag_Missing
df["Flag_Missing"] = df.isnull().any(axis=1).astype(int)

# 2. Tỷ lệ dòng có missing
ty_le_missing = df["Flag_Missing"].mean() * 100
print("Tỷ lệ dòng có missing: {:.2f}%".format(ty_le_missing))

# 3. So sánh doanh thu trung bình
doanhthu_tb = df.groupby("Flag_Missing")["Thành tiền (VNĐ)"].mean()

print("\nDoanh thu trung bình:")
print("Không missing (0):", doanhthu_tb[0])
print("Có missing (1):", doanhthu_tb[1])

# 4. Nhận xét tự động
if doanhthu_tb[1] > doanhthu_tb[0]:
    print("\nNhận xét: Nhóm có missing có doanh thu trung bình CAO hơn → có thể dữ liệu thiếu xuất hiện ở đơn hàng lớn.")
elif doanhthu_tb[1] < doanhthu_tb[0]:
    print("\nNhận xét: Nhóm có missing có doanh thu trung bình THẤP hơn → có thể dữ liệu thiếu xuất hiện ở đơn hàng nhỏ.")
else:
    print("\nNhận xét: Hai nhóm có doanh thu tương đương → missing không ảnh hưởng đáng kể.")
```

(A) Drop tất cả dòng có missing, 
(B) Điền tất cả bằng median/mode, 
(C) Điền theo nhóm. So sánh: kích thước, mean Thành tiền, std Thành tiền của 3 phương án. 

Phương án nào bạn khuyến nghị?

```python
# Câu 4.10
df_copy = df.copy()

# Phương án A
df_A = df_copy.dropna()
# Phương án A – Drop
# Kích thước giảm mạnh
# Mean và std có thể bị lệch (do mất dữ liệu)
# Mất thông tin

# Phương án B
df_B = df_copy.copy()
# Điền số -> median
for col in df_B.select_dtypes(include=['number']).columns:
    df_B[col].fillna(df_B[col].median(), inplace=True)
# Phương án B – Fill median/mode
# Giữ nguyên số dòng
# Mean ổn định hơn
# Std giảm (do dữ liệu bị “làm phẳng”)
# Đơn giản nhưng thiếu chính xác theo ngữ cảnh
    
# Điền categorical -> mode
for col in df_B.select_dtypes(include=['object']).columns:
    df_B[col].fillna(df_B[col].mode()[0], inplace=True)
    
# Phương án C
df_C = df_copy.copy()

df_C["Thành tiền (VNĐ)"] = df_C.groupby("Trạng thái")["Thành tiền (VNĐ)"].transform(lambda x: x.fillna(x.median()))
df_C["Đánh giá (sao)"] = df_C.groupby("Trạng thái")["Đánh giá (sao)"].transform(lambda x: x.fillna(x.median()))
# Phương án C – Fill theo nhóm
# Giữ nguyên dữ liệu
# Mean hợp lý
# Std gần với thực tế hơn
# Phản ánh đúng bản chất dữ liệu hơn

# So sánh 3 chiến lươc
def summary(df):
    return {
        "size": len(df),
        "mean": df["Thành tiền (VNĐ)"].mean(),
        "std": df["Thành tiền (VNĐ)"].std()
    }


result = {
    "A_drop": summary(df_A),
    "B_fill": summary(df_B),
    "C_group": summary(df_C)
                     
}

pd.DataFrame(result)

# Nhận xét
# Phương án (A) làm giảm đáng kể kích thước dữ liệu, dẫn đến mất thông tin và có thể gây sai lệch kết quả phân tích.
# Phương án (B) giữ nguyên dữ liệu nhưng làm giảm độ biến thiên do sử dụng giá trị trung vị/chế độ chung cho toàn bộ dữ liệu.
# Phương án (C) là hợp lý nhất vì tận dụng thông tin theo từng nhóm, giúp bảo toàn đặc trưng phân phối dữ liệu tốt hơn.

# ⇒ Khuyến nghị sử dụng phương án (C) – điền missing theo nhóm.
```

### Câu 5.1

Dùng `groupby` tính **số đơn hàng** và **tổng `Thành tiền (VNĐ)`** theo từng `Khu vực`. Sắp xếp theo tổng doanh thu giảm dần.

```python
# Câu 5.1
result = df.groupby("Khu vực").agg(
    soDonHang = ("Khu vực", "count"),
    tongThanhTien = ("Thành tiền (VNĐ)", "sum")
).sort_values("tongThanhTien",ascending=False)

print(result)
```

### Câu 5.2

Tính **doanh thu trung bình, trung vị, độ lệch chuẩn** của `Thành tiền (VNĐ)` theo từng `Nhóm sản phẩm`. Nhóm nào có doanh thu trung bình cao nhất? Nhóm nào ổn định nhất (std thấp nhất)?

```python
# Câu 5.2
result = df.groupby("Nhóm sản phẩm").agg(
    doanhThuTB = ("Thành tiền (VNĐ)", "sum"),
    median = ("Thành tiền (VNĐ)", "median"),
    std = ("Thành tiền (VNĐ)", "std")
)
print(result)
print("\nNhóm có doanh thu trung bình cao nhất là:",result["doanhThuTB"].idxmax(),"(",result["doanhThuTB"].max(),")" )
print("\nNhóm ổn định nhất là:",result["std"].idxmin(),"(", result["std"].min(),")")
```

### Câu 5.3

Tính **đánh giá trung bình** theo từng `Kênh bán hàng`. Kênh nào được đánh giá cao nhất?

```python
# Câu 5.3
result = df.groupby("Kênh bán hàng").agg(
    danhGiaTB = ("Đánh giá (sao)", "mean")
)

print(result)
print("Kênh được đánh giá cao nhất là:",result["danhGiaTB"].idxmax(), "(", round(result["danhGiaTB"].max(), 2), "sao)")
```

### Câu 5.4

Gom nhóm theo **tháng** (từ `Ngày đặt hàng`), tính **tổng doanh thu** và **số đơn hàng** theo từng tháng. Tháng nào có doanh thu cao nhất?

```python
# Câu 5.4
dateFromMonth = df["Ngày đặt hàng"].dt.month
result = df.groupby(dateFromMonth).agg(
    tongDoanhThu = ("Doanh thu (VNĐ)", "sum"),
    soDon = ("Mã đơn hàng", "count")
)

print(result)

print("Tháng có doanh thu cao nhất là tháng", result["tongDoanhThu"].idxmax())


```

### Câu 5.9

Phân tích **RFM đơn giản** cho mỗi `Mã nhân viên`: (R) Ngày bán hàng gần nhất, (F) Tổng số đơn, (M) Tổng doanh thu. 

Xếp hạng top 10 nhân viên theo từng tiêu chí.

Ghi chú: Có thể đọc thêm về phân tích RFM

```python
# Câu 5.9
rfm = df.groupby("Mã nhân viên").agg(
    R = ("Ngày đặt hàng", "max"),
    F = ("Mã đơn hàng", "count"),
    M = ("Doanh thu (VNĐ)", "sum")
)

print(rfm)

top_R = rfm["R"].sort_values(ascending=False).head(10)
print(top_R)

top_F = rfm["F"].sort_values(ascending=False).head(10)
print(top_F)

top_M = rfm["M"].sort_values(ascending=False).head(10)
print(top_M)
```

### Câu 5.10

Tạo **Pivot Table lồng ghép** (multi-index): dòng = (`Khu vực`, `Nhóm sản phẩm`), cột = `Quý` (Q1–Q4 của 2024), giá trị = tổng `Thành tiền (VNĐ)`. Tính thêm tăng trưởng % giữa các quý cho mỗi (Khu vực × Nhóm SP).

```python
# Câu 5.10

df["Ngày đặt hàng"] = pd.to_datetime(df["Ngày đặt hàng"])

df["Năm"] = df["Ngày đặt hàng"].dt.year
df["Quý"] = df["Ngày đặt hàng"].dt.quarter

df_2024 = df[df["Năm"]==2024]

pivot = pd.pivot_table(
    df_2024,
    index= ["Khu vực", "Nhóm sản phẩm"],
    columns= "Quý",
    values= "Thành tiền (VNĐ)",
    aggfunc= "sum",
    fill_value=0
)

print(pivot)

growth = pivot.pct_change(axis=1)*100
print(growth)


# Pivot Table cho thấy doanh thu theo từng Khu vực và Nhóm sản phẩm theo quý trong năm 2024.
# Tốc độ tăng trưởng giữa các quý giúp đánh giá xu hướng kinh doanh, xác định các nhóm sản phẩm và khu vực có sự tăng trưởng hoặc suy giảm theo thời gian.
```

```python
# Câu 6.10
plt.figure(figsize=(12,6))

# Violot plot split theo giới tính
sns.violinplot(
    data=df,
    x="Nhóm sản phẩm",
    y="Tuổi KH",
    hue="Giới tính KH",
    split=True,
    palette="Set2",
    inner=None
)

# Thêm điểm mean

means = df.groupby(["Nhóm sản phẩm", "Giới tính KH"])["Tuổi KH"].mean().reset_index()

# Lấy vị trí nhóm để đặt điểm
groups = df["Nhóm sản phẩm"].unique()

for i, group in enumerate(groups):
    for gender in df["Giới tính KH"].unique():
        m = means[
            (means["Nhóm sản phẩm"] == group) &
            (means["Giới tính KH"] == gender)
        ]["Tuổi KH"]
        
        if not m.empty:
            x = i - 0.2 if gender == df["Giới tính KH"].unique()[0] else i + 0.2
            plt.scatter(x, m.values[0], color="black", s=60, zorder=5)
            
plt.title("Phân phối tuổi khách hàng theo nhóm sản phẩm và giới tính")
plt.xticks(rotation=30)
plt.show()
```

# CHỦ ĐỀ 7: XÁC SUẤT, XÁC SUẤT CÓ ĐIỀU KIỆN, ĐỊNH LÝ BAYES
> **Hướng dẫn:** Sử dụng tần suất từ dữ liệu thực tế để **ước lượng xác suất** (relative frequency approach).
> - P(A) = số dòng thỏa A / tổng số dòng
> - P(A|B) = số dòng thỏa A∩B / số dòng thỏa B

### Câu 7.1

Tính các xác suất sau (dựa trên tần suất tương đối):
- P(Khu vực = "Miền Nam")
- P(Trạng thái = "Đã hủy")
- P(Nhóm sản phẩm = "Điện tử")
- P(Giảm giá > 0)

```python
# Câu 7.1
# Tổng số dòng
n = len(df)
# 1. P(Khu vực = "Miền Nam")
p_MienNam = (df["Khu vực"] == "Miền Nam").sum()/n
p_huy = (df["Trạng thái"] == "Đã hủy").sum()/n
p_nhomSP = (df["Nhóm sản phẩm"] == "Điện tử").sum()/n
p_giamgia = (df["Giảm giá (%)"] > 1).sum()/n
print("P(Khu vực = Miền Nam) =",round(p_MienNam,2))
print("P(Trạng thái = Đã hủy) =",round(p_huy,2))
print("P(Nhóm sản phẩm = Điện tử) =",round(p_nhomSP,2))
print("P(Giảm giá > 0) =",round(p_giamgia,2))
```

### Câu 7.2

Tính **xác suất có điều kiện**:
- P(Hoàn thành | Miền Nam) — xác suất đơn hoàn thành nếu biết đơn ở Miền Nam
- P(Hoàn thành | Tây Nguyên)
- So sánh hai xác suất này. Khu vực có ảnh hưởng đến tỷ lệ hoàn thành không?

```python
# Câu 7.2

# Cách 1: Do tôi tự làm
getData = df.groupby("Khu vực")["Trạng thái"].value_counts()

miennam = getData["Miền Nam"]
totalDonHangMienNam = miennam.sum()
soDonHoanThanhMienNam = miennam["Hoàn thành"]
p_miennam = soDonHoanThanhMienNam / totalDonHangMienNam

taynguyen = getData["Tây Nguyên"]
totalDonHangTayNguyen = taynguyen.sum()
soDonHoanThanhTayNguyen = taynguyen["Hoàn thành"]
p_taynguyen = soDonHoanThanhTayNguyen / totalDonHangTayNguyen

print("P(Hoàn thành | Miền Nam) =",round(p_miennam,2))
print("P(Hoàn thành | Tây Nguyên) =",round(p_taynguyen,2))

# Cách 2: Do AI chỉ
p_miennam = (
    df[df["Khu vực"] == "Miền Nam"]["Trạng thái"]
    .eq("Hoàn thành")
    .mean()
)

p_taynguyen = (
    df[df["Khu vực"] == "Tây Nguyên"]["Trạng thái"]
    .eq("Hoàn thành")
    .mean()
)

print("P(Hoàn thành | Miền Nam) =", round(p_miennam, 2))
print("P(Hoàn thành | Tây Nguyên) =", round(p_taynguyen, 2))
# Nhận xét
# Ta thấy tỷ lệ thành công ở miền nam cao hơn ở tây nguyên
# Chênh lệch là 0,08 = 8% khá rỏ ràng
# Vậy khu vực cũng ảnh hưởng đến tỷ lệ hoàn thành của đơn hàng
```

### Câu 7.3

Tính:
- P(Nữ ∩ Mỹ phẩm) — xác suất đồng thời
- P(Nữ) × P(Mỹ phẩm) — tích xác suất riêng lẻ
- So sánh hai giá trị: sự kiện "Nữ" và "mua Mỹ phẩm" có **độc lập** không?

```python
# Câu 7.3
p_nu = (df["Giới tính KH"] == "Nữ").mean()
p_mypham = (df["Nhóm sản phẩm"] == "Mỹ phẩm").mean()

p_nu_mypham = ((df["Giới tính KH"] == "Nữ") & (df["Nhóm sản phẩm"] == "Mỹ phẩm")).mean()

print("P(Nữ ∩ Mỹ phẩm) — xác suất đồng thời =", round(p_nu_mypham,4))
print("P(Nữ) × P(Mỹ phẩm) — xác suất riêng lẻ =", round(p_nu*p_mypham,4))
if abs(p_nu_mypham - p_nu*p_mypham) < 0.01:
    print("Hai biến cố có độc lập")
else:
    print("Hai biến cố không độc lập")
```

### Câu 7.4

Tính xác suất **hợp (union)**:
- P(Đã hủy ∪ Trả hàng) — xác suất đơn hàng bị hủy HOẶC trả hàng
- Kiểm tra: P(A ∪ B) = P(A) + P(B) − P(A ∩ B) có thỏa mãn không? (A = Đã hủy, B = Trả hàng)

```python
# Câu 7.4
p_A_or_B = ((df["Trạng thái"] == "Đã hủy") | (df["Trạng thái"] == "Trả hàng")).mean()
print("P(Đã hủy ∪ Trả hàng) =",p_A_or_B)
p_A = (df["Trạng thái"] == "Đã hủy").mean()
p_B = (df["Trạng thái"] =="Trả hàng").mean()
print("P(Đã hủy) =",p_A)
print("P(Trả hàng) =",p_B)
p_A_and_B = ((df["Trạng thái"] == "Đã hủy") & (df["Trạng thái"] == "Trả hàng")).mean()
print("P(A ∩ B) =",p_A_and_B)
check = p_A + p_B - p_A_and_B
if abs(p_A_or_B - check) < 1e-9:
    print("P(A ∪ B) = P(A) + P(B) − P(A ∩ B) thỏa mãn")
else:
    print("P(A ∪ B) = P(A) + P(B) − P(A ∩ B) không thỏa mãn")
    
```

### Câu 7.5

Một đơn hàng được chọn ngẫu nhiên. Tính:
- P(Thành tiền > 1 triệu)
- P(Thành tiền > 5 triệu)
- P(Thành tiền > 5 triệu | Nhóm SP = "Điện tử")
- Nhận xét: biết nhóm SP là Điện tử làm thay đổi xác suất như thế nào?

```python
# Câu 7.5
p_A = (df["Thành tiền (VNĐ)"] > 1000000).mean()
print(p_A)

p_B = (df["Thành tiền (VNĐ)"] > 5000000).mean()
print(p_B)

p_dientu = (df["Nhóm sản phẩm"] == "Điện tử").mean()
p_C = ((df["Thành tiền (VNĐ)"] > 5000000)&(df["Nhóm sản phẩm"] == "Điện tử")).mean()
p_cond = p_C / p_dientu
print(p_cond)

# Nhận xét
# Biết đơn hàng thuộc nhóm Điện tử làm tăng đáng kể xác suất giá trị đơn hàng lớn (>5 triệu), từ 15.8% lên 55.2%. 
# Điều này cho thấy nhóm sản phẩm có ảnh hưởng mạnh đến hành vi chi tiêu của khách hàng,
# đặc biệt nhóm Điện tử có xu hướng tạo ra các đơn hàng giá trị cao.
```

### Câu 7.6

Tính xác suất có điều kiện theo **kênh bán hàng**:
- P(Đánh giá ≥ 4 sao | Online)
- P(Đánh giá ≥ 4 sao | Cửa hàng)
- P(Đánh giá ≥ 4 sao | Đại lý)
- Kênh nào có xác suất khách hàng hài lòng cao nhất?

```python
# Câu 7.6
p_onl = (df["Kênh bán hàng"]=="Online").mean()
p_4sao_onl = ((df["Đánh giá (sao)"] >= 4) & (df["Kênh bán hàng"]=="Online")).mean()
p_xsA = p_4sao_onl / p_onl  
print(round(p_xsA,4))

p_cuahang = (df["Kênh bán hàng"]=="Cửa hàng").mean()
p_4sao_cuahang = ((df["Đánh giá (sao)"] >= 4) & (df["Kênh bán hàng"]=="Cửa hàng")).mean()
p_xsB = p_4sao_cuahang / p_cuahang
print(round(p_xsB,4))

p_daily = (df["Kênh bán hàng"]=="Đại lý").mean()
p_4sao_daily = ((df["Đánh giá (sao)"] >= 4) & (df["Kênh bán hàng"]=="Đại lý")).mean()
p_xsC = p_4sao_daily / p_daily
print(round(p_xsC,4))

if(p_xsA > p_xsB and p_xsA > p_xsC):
    print("Kênh bán hàng Online có xác xuất khách hàng hài lòng cao nhất")
elif(p_xsB > p_xsC):
    print("Kênh bán hàng Cửa Hàng có xác xuất khách hàng hài lòng cao nhất")
else:
    print("Kênh bán hàng Đại lý có xác xuất khách hàng hài lòng cao nhất")
```

### Câu 7.7

Tính:
- P(Nam | Điện tử) — xác suất KH là nam biết mua Điện tử
- P(Nam | Mỹ phẩm) — xác suất KH là nam biết mua Mỹ phẩm
- P(Điện tử | Nam) — xác suất mua Điện tử biết KH là nam
- Giải thích sự khác biệt giữa P(A|B) và P(B|A).

```python
# Câu 7.7
p_dientu = (df["Nhóm sản phẩm"]=="Điện tử").mean()
p_namdientu = ((df["Giới tính KH"] == "Nam") & (df["Nhóm sản phẩm"]=="Điện tử")).mean()
p_xsA = p_namdientu / p_dientu  
print(round(p_xsA,4))

p_mypham = (df["Nhóm sản phẩm"]=="Mỹ phẩm").mean()
p_nammypham = ((df["Giới tính KH"] == "Nam") & (df["Nhóm sản phẩm"]=="Mỹ phẩm")).mean()
p_xsB = p_nammypham / p_mypham
print(round(p_xsB,4))

p_nam = (df["Giới tính KH"] == "Nam").mean()
p_dientunam = ((df["Nhóm sản phẩm"]=="Điện tử") & (df["Giới tính KH"] == "Nam")).mean()
p_xsC = p_dientunam / p_nam
print(round(p_xsC,4))

# Nhận xét
# Xác suất có điều kiện phụ thuộc vào biến được chọn làm điều kiện.
# P(Nam | Điện tử) và P(Điện tử | Nam) cho hai góc nhìn khác nhau:
# một bên xét tỷ lệ giới tính trong nhóm sản phẩm, bên còn lại xét hành vi mua hàng của một giới tính.
# Do đó, hai xác suất này thường không bằng nhau và không thể thay thế cho nhau.
```

### Câu 7.8

**Ứng dụng phép cộng xác suất:**
Chọn ngẫu nhiên 1 đơn hàng. Tính xác suất đơn hàng đó:
- Thuộc Miền Nam **VÀ** mua Điện tử **VÀ** thanh toán bằng Ví điện tử
- Thuộc Miền Nam **HOẶC** mua Điện tử (dùng công thức P(A∪B))

```python
# Câu 7.8
p_donhangA = (
    (df["Khu vực"] == "Miền Nam") & 
    (df["Nhóm sản phẩm"] == "Điện tử") & 
    (df["Phương thức thanh toán"] == "Ví điện tử")
).mean()

# Cách 1
# P(A∪B)=P(A)+P(B)−P(A∩B)
p_A = (df["Khu vực"] == "Miền Nam").mean()
p_B = (df["Nhóm sản phẩm"] == "Điện tử").mean()
p_A_and_B = (
    (df["Khu vực"] == "Miền Nam") &
    (df["Nhóm sản phẩm"] == "Điện tử") 
).mean()

p_donhangB = p_A + p_B - p_A_and_B
# Cách 2
# p_donhangB = (
#     (df["Khu vực"] == "Miền Nam") |
#     (df["Nhóm sản phẩm"] == "Điện tử") 
# ).mean()

print(p_donhangA)
print(p_donhangB)
```

### Câu 7.9

Định lý Bayes — Bài toán chẩn đoán:**

Biết rằng:
- Đơn hàng bị hủy (H) chiếm tỷ lệ P(H) trong dữ liệu
- Xác suất khách hàng trẻ (≤25 tuổi) trong nhóm bị hủy: P(Trẻ|H)
- Xác suất khách hàng trẻ (≤25 tuổi) trong nhóm KHÔNG bị hủy: P(Trẻ|H̄)

**Hãy tính:** Nếu biết một khách hàng ≤ 25 tuổi, xác suất đơn hàng đó bị hủy là bao nhiêu?
→ Áp dụng: **P(H|Trẻ) = P(Trẻ|H) × P(H) / P(Trẻ)**

Tính P(Trẻ) bằng công thức xác suất toàn phần: P(Trẻ) = P(Trẻ|H)×P(H) + P(Trẻ|H̄)×P(H̄)

```python
# Câu 7.9

# B1 tích xác suất từng phần
# P(H)
p_H = (df["Trạng thái"] == "Đã hủy").mean()

# P(Not H)
p_not_H = 1 - p_H

# P(Trẻ|H)
df_H = df[df["Trạng thái"] == "Đã hủy"]
p_tre_given_H = (df_H["Tuổi KH"] <= 25).mean()

# P(trẻ | not H)
df_not_H = df[df["Trạng thái"] != "Đã hủy"]
p_tre_given_not_H = (df_not_H["Tuổi KH"] <= 25).mean()

# B2 Tính P trẻ
p_tre = p_tre_given_H*p_H + p_tre_given_not_H*p_not_H

# B3 Áp dụng Bayes
p_H_given_tre = (p_tre_given_H*p_H) / p_tre

print("P(H|Trẻ) =",round(p_H_given_tre,4))
```

### Câu 7.10

Định lý Bayes — Bài toán phân loại kênh bán hàng:**

Một đơn hàng vừa được ghi nhận có: `Nhóm sản phẩm = "Điện tử"` và `Trạng thái = "Hoàn thành"`.

Sử dụng Naive Bayes đơn giản, hãy tính xác suất đơn hàng này thuộc **từng kênh bán hàng** (Cửa hàng, Online, Đại lý, Siêu thị):

P(Kênh = k | Điện tử ∩ Hoàn thành) ∝ P(Điện tử | k) × P(Hoàn thành | k) × P(k)

Tính posterior cho mỗi kênh, chuẩn hóa để tổng = 1. Kênh nào có xác suất cao nhất?

```python
# Câu 7.10
channels = ["Cửa hàng", "Online", "Đại lý", "Siêu thị"]

results = {}

for k in channels:
    df_k = df[df["Kênh bán hàng"] == k]
    
#     P(k)
    p_k = len(df_k) / len(df)
    
#     P(Điện tử / k)
    p_dt_given_k = (df_k["Nhóm sản phẩm"] == "Điện tử").mean()
    
#     P(Hoàn thành / k)
    p_ht_given_k = (df_k["Trạng thái"] == "Hoàn thành").mean()
    
#     Posterior chưa chuẩn hóa
    posterior = p_dt_given_k*p_ht_given_k*p_k
    results[k] = posterior
    
# CHuẩn hóa
total = sum(results.values())
for k in results:
    results[k] = results[k] / total
print(results)
```
