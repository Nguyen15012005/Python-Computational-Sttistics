### Câu 2.4

Tính **Q1, Q2 (median), Q3, IQR** của cột `Số lượng`. Xác định ngưỡng outlier trên và dưới theo phương pháp IQR.

```python
# Câu 2.4
Q1 = df["Số lượng"].quantile(0.25)
Q2 = df["Số lượng"].median()
Q3 = df["Số lượng"].quantile(0.75)

IQR = Q3 - Q1

khoangduoi = Q1 - 1.5*IQR
khoangtren = Q3 + 1.5*IQR

outliers = df[(df["Số lượng"] < khoangduoi) | (df["Số lượng"] > khoangtren)]
print(outliers)
```

### Câu 2.10

Tính **Percentile thứ 5, 10, 25, 50, 75, 90, 95** cho `Thành tiền (VNĐ)`. 

Dựa vào kết quả, hãy phân nhóm đơn hàng thành 4 nhóm: Nhỏ (≤P25), Trung bình (P25–P75), Lớn (P75–P95), Rất lớn (>P95). 

Đếm số đơn hàng và tính tổng doanh thu mỗi nhóm.

```python
# Câu 2.10
percentiles = df["Thành tiền (VNĐ)"].quantile([0.05, 0.1, 0.25, 0.5, 0.75, 0.9, 0.95])
print(percentiles)
p_25 = percentiles.loc[0.25]
print(p_25)
p_75 = percentiles.loc[0.75]
print(p_75)
p_95 = percentiles.loc[0.95]
print(p_95)
def classify(x):
    if x <= p_25:
        return "Nhỏ"
    elif x <= p_75:
        return "Trung bình"
    elif x <= p_95:
        return "Lớn"
    else:
        return "rất lớn"
    
df["Nhóm đơn hàng"] = df["Thành tiền (VNĐ)"].apply(classify)

result = df.groupby("Nhóm đơn hàng").agg(
    so_don=("Mã đơn hàng", "count"),
    tong_doanh_thu = ("Doanh thu (VNĐ)", "sum")
)

print(result)
```

### Câu 3.8

Sử dụng phương pháp **IQR** để tìm outlier trong cột `Đơn giá (VNĐ)`. Các outlier chủ yếu thuộc nhóm sản phẩm nào?

```python
# Câu 3.8
Q1 = df["Đơn giá (VNĐ)"].quantile(0.25)
Q2 = df["Đơn giá (VNĐ)"].median()
Q3 = df["Đơn giá (VNĐ)"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - IQR*1.5
upper_bound = Q3 + IQR*1.5

result = df[
    (df["Đơn giá (VNĐ)"] < lower_bound) |
    (df["Đơn giá (VNĐ)"] > upper_bound) 
]
print(result["Nhóm sản phẩm"].value_counts())
print("Các outlier chủ yếu thuộc nhóm sản phẩm: ",result["Nhóm sản phẩm"].value_counts().idxmax())
```

### Câu 3.9

Tìm **top 3 nhân viên** (theo `Mã nhân viên`) có tổng doanh thu (`Thành tiền`) cao nhất. Với mỗi nhân viên, cho biết: số đơn hàng, doanh thu trung bình/đơn, nhóm sản phẩm bán nhiều nhất, tỷ lệ đơn hoàn thành.

```python
# Câu 3.9
result = df.groupby("Mã nhân viên")["Thành tiền (VNĐ)"].sum().sort_values(ascending=False).head(3)
print("3 nhân viên có tổng thành tiền cao nhất: ")
print(result)
listnvtop3 = result.index

df_top3 = df[df["Mã nhân viên"].isin(listnvtop3)]

result = []
for nv in listnvtop3:
    data_top3 = df_top3[df_top3["Mã nhân viên"] == nv]
#     Số đơn
    so_don = len(data_top3)
#   doanh thu trung bình
    doanhthutb = data_top3["Thành tiền (VNĐ)"].mean()
#   Nhóm sản phẩm bán nhiều nhất
    nhomSp = data_top3["Nhóm sản phẩm"].value_counts().idxmax()
#   Tỷ lệ hoàn thành
    tyle = (data_top3["Trạng thái"] == "Hoàn thành").mean()*100
    result.append({
        "Mã nhân viên":nv,
        "Số đơn": so_don,
        "Doanh thu trung bình": doanhthutb,
        "Nhóm sản phẩm bán chạy nhất": nhomSp,
        "Tỷ lệ đơn hoàn thành" : tyle
        
    })
res = pd.DataFrame(result)
print(res)
```

### Câu 4.1

```python
# Câu 4.1
missing = df.isnull().sum()
tyle = missing / len(df) * 100

bangmissing = pd.DataFrame({
    "Tên cột": missing.index,
    "Số lượng missing": missing.values,
    "Tỷ lệ %": tyle.values
})

bangmissing = bangmissing[bangmissing["Số lượng missing"] > 0]
bangmissing = bangmissing.sort_values("Tỷ lệ %", ascending=False)
print(bangmissing)
```

### Câu 4.2

Tạo biểu đồ thanh (bar chart) thể hiện **tỷ lệ missing** của các cột có missing. Cột nào cần ưu tiên xử lý?

### Câu 4.3

Với cột `Tuổi KH`, hãy điền giá trị missing bằng **median**. So sánh mean và std trước và sau khi điền. Giá trị thay đổi nhiều không?

```python
# Câu 4.3
df["Tuổi KH"] = df["Tuổi KH"].astype("float64")
print("Median trước:", df["Tuổi KH"].mean())
print("Std trước:", df["Tuổi KH"].std())
median = df["Tuổi KH"].median()
df["Tuổi KH"].fillna(median, inplace=True)
print("Median trước:", df["Tuổi KH"].mean())
print("Std trước:", df["Tuổi KH"].std())

```

### Câu 4.4

Với cột `Đánh giá (sao)`, hãy điền missing bằng **mode** (giá trị xuất hiện nhiều nhất). Giải thích tại sao mode phù hợp hơn mean cho biến này?

```python
# Câu 4.4
mode = df["Đánh giá (sao)"].mode()
df["Đánh giá (sao)"].fillna(mode, inplace=True)
```

### Câu 4.5

Với cột `Phí vận chuyển (VNĐ)`, hãy điền missing bằng **mean theo nhóm `Khu vực`** (mỗi khu vực có mean khác nhau). Tại sao cách này tốt hơn điền mean chung?

```python
# Câu 4.5
df["Phí vận chuyển (VNĐ)"] = df["Phí vận chuyển (VNĐ)"].fillna(df.groupby("Khu vực")["Phí vận chuyển (VNĐ)"].transform("mean")) 
df["Phí vận chuyển (VNĐ)"]
```

### Câu 4.6

Với cột `Giới tính KH`, hãy điền missing bằng phương pháp **random sampling** (lấy ngẫu nhiên từ các giá trị không missing, giữ nguyên tỷ lệ Nam/Nữ). Kiểm tra tỷ lệ Nam/Nữ trước và sau.

```python
# Câu 4.6
print("Tỷ lệ trước")
print(df["Giới tính KH"].value_counts(normalize=True))

missing = df["Giới tính KH"].isnull()
no_missing = df["Giới tính KH"].dropna()
df.loc[missing, "Giới tính KH"] = np.random.choice(no_missing, size=missing.sum())
print("Tỷ lệ sau:")
print(df["Giới tính KH"].value_counts(normalize=True))
```

# CHỦ ĐỀ 5: GOM NHÓM, TỔNG HỢP DỮ LIỆU, PIVOT TABLE

### Câu 5.1

Dùng `groupby` tính **số đơn hàng** và **tổng `Thành tiền (VNĐ)`** theo từng `Khu vực`. Sắp xếp theo tổng doanh thu giảm dần.

```python
# Câu 5.1
result = df.groupby("Khu vực").agg(
    sodonhang = ("Mã đơn hàng", "count"),
    tongthanhtien =("Thành tiền (VNĐ)", "sum"),
    tongdoanhthu =("Doanh thu (VNĐ)", "sum")
).sort_values(by="tongdoanhthu", ascending=False)

result
```

### Câu 5.2

Tính **doanh thu trung bình, trung vị, độ lệch chuẩn** của `Thành tiền (VNĐ)` theo từng `Nhóm sản phẩm`. Nhóm nào có doanh thu trung bình cao nhất? Nhóm nào ổn định nhất (std thấp nhất)?

```python
# Câu 5.2
# ổn định nhất => std min
result = df.groupby("Nhóm sản phẩm").agg(
    doanhthuTB = ("Thành tiền (VNĐ)", "mean"),
    trungvi = ("Thành tiền (VNĐ)", "median"),
    std = ("Thành tiền (VNĐ)", "std")
)
print(result)
print("Nhóm có doanh thu trung bình cao nhất:", result["doanhthuTB"].idxmax())
print("Nhóm ổn định nhất:", result["std"].idxmin())
```

### Câu 5.3

Tính **đánh giá trung bình** theo từng `Kênh bán hàng`. Kênh nào được đánh giá cao nhất?

```python
# Câu 5.3
result = df.groupby("Kênh bán hàng")["Đánh giá (sao)"].mean()
print(result)

print("Kênh được đánh giá cao nhất:", result.idxmax())
```

### Câu 5.4

Gom nhóm theo **tháng** (từ `Ngày đặt hàng`), tính **tổng doanh thu** và **số đơn hàng** theo từng tháng. Tháng nào có doanh thu cao nhất?

```python
# Câu 5.4
df["Tháng"] = df["Ngày đặt hàng"].dt.month
result = df.groupby("Tháng").agg(
    tongdoanhthu = ("Doanh thu (VNĐ)", "sum"),
    sodonhang = ("Mã đơn hàng", "count")
)

print(result)
print("Tháng có doanh thu cao nhất là tháng", result["tongdoanhthu"].idxmax())
```

### Câu 5.5

Tạo **Pivot Table**: dòng = `Khu vực`, cột = `Nhóm sản phẩm`, giá trị = tổng `Thành tiền (VNĐ)`. Thêm tổng dòng và tổng cột.

```python
# Câu 5.5

result = pd.pivot_table(
    df,
    index= "Khu vực",
    columns= "Nhóm sản phẩm",
    values="Thành tiền (VNĐ)",
    aggfunc="sum",
    margins=True,
    margins_name="Tổng"
)
result
```

### Câu 5.6

Tạo **Pivot Table**: dòng = `Phương thức thanh toán`, cột = `Trạng thái`, giá trị = **đếm** số đơn hàng. Phương thức nào có tỷ lệ hủy cao nhất?

```python
# Câu 5.6
result = pd.pivot_table(
    df,
    index= "Phương thức thanh toán",
    columns= "Trạng thái",
    values="Mã đơn hàng",
    aggfunc="count",
    margins=True,
    margins_name="Tổng"
)

result["Tỷ lệ hủy %"] = result["Đã hủy"] / result["Tổng"]*100

print(result)
print("Phương thức có tỷ lệ hủy cao nhất:",result["Tỷ lệ hủy %"].idxmax())
```

### Câu 5.7

Gom nhóm theo **Khu vực × Kênh bán hàng** (2 cấp). Tính tổng doanh thu và đếm số đơn. Sự kết hợp nào mang lại doanh thu cao nhất?

```python
# Câu 5.7
result = df.groupby(["Khu vực", "Kênh bán hàng"]).agg(
    tongdoanhthu = ("Doanh thu (VNĐ)", "sum"),
    sodon = ("Mã đơn hàng", "count")
)
print(result)
print("Sự kết hợp mang lại doanh thu cao nhất:",result["tongdoanhthu"].idxmax())
```

Tạo cột `Nhóm tuổi`: 18–25, 26–35, 36–45, 46–55, 56+. Tính **doanh thu trung bình** và **sản phẩm mua nhiều nhất** theo từng nhóm tuổi.

### Câu 5.9

Phân tích **RFM đơn giản** cho mỗi `Mã nhân viên`: (R) Ngày bán hàng gần nhất, (F) Tổng số đơn, (M) Tổng doanh thu. 

Xếp hạng top 10 nhân viên theo từng tiêu chí.

Ghi chú: Có thể đọc thêm về phân tích RFM

```python
# Câu 5.9
rfm = df.groupby("Mã nhân viên").agg(
    r = ("Ngày đặt hàng", "max"),
    f = ("Mã đơn hàng", "count"),
    m = ("Doanh thu (VNĐ)", "sum")
)

Top10R = rfm["r"].sort_values(ascending=False).head(10)
print(Top10R)

Top10F = rfm["f"].sort_values(ascending=False).head(10)
print(Top10F)

Top10M = rfm["m"].sort_values(ascending=False).head(10)
print(Top10M)
```

### Câu 5.10

Tạo **Pivot Table lồng ghép** (multi-index): dòng = (`Khu vực`, `Nhóm sản phẩm`), cột = `Quý` (Q1–Q4 của 2024), giá trị = tổng `Thành tiền (VNĐ)`. Tính thêm tăng trưởng % giữa các quý cho mỗi (Khu vực × Nhóm SP).

```python
# Câu 5.10
result = pd.pivot_table(
    df[df["Ngày đặt hàng"].dt.year == 2024],
    index=["Khu vực", "Nhóm sản phẩm"],
    columns=(df["Ngày đặt hàng"].dt.quarter),
    values = "Thành tiền (VNĐ)",
    aggfunc="sum",
    fill_value=0
)

print(result)
growth = result.pct_change(axis=1)*100
print(growth)
```
