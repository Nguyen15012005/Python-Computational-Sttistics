```python
# Chuẩn bị môi trường
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['figure.figsize'] = (10, 5)
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3

# Đọc dữ liệu
df_orders = pd.read_csv('order_values.csv')
df_delivery = pd.read_csv('delivery_times.csv')
df_ab = pd.read_csv('ab_test_conversions.csv')

print('=== Dataset 1: order_values.csv ===')
print(df_orders.head())
print(f'\nSố dòng: {len(df_orders)}')
print(df_orders.groupby('group').describe()['order_value_vnd'].round(0))
```

```python
print('=== Dataset 2: delivery_times.csv ===')
print(df_delivery.head())
print(f'\nSố dòng: {len(df_delivery)}')
print(df_delivery.groupby('carrier').describe()['delivery_minutes'].round(2))
```

```python
print('=== Dataset 3: ab_test_conversions.csv ===')
print(df_ab.head())
print(f'\nSố dòng: {len(df_ab)}')
print(df_ab.groupby('variant').agg(
    n_sessions=('session_id', 'count'),
    n_converted=('converted', 'sum'),
    conversion_rate=('converted', 'mean')
).round(4))
```
